### 2. 🛣️🎯 Choosing Your Path: How Python & Mojo Connect

While Mojo and Python share a soul 💫, they can work together in different ways depending on your performance needs:

* **🔄 Direct Interop (Python ↔ Mojo):** Call Python libraries directly from Mojo or wrap Mojo functions for Python—perfect for quick integration and "Mojo-fying" existing scripts. ⚡🔧
* **📊 MAX Graph (The Optimization Engine):** Construct a symbolic execution plan in Python and "Register" Mojo kernels as Custom Ops—this is how you achieve peak hardware utilization and model-wide fusion. 🎯🚀

### 3. 🌟💎 Why MAX Graph is the Star of the Show Today

We are focusing on **MAX Graph** because it represents the highest level of performance engineering:

* **⏰ Ahead-of-Time (AOT) Planning:** The Graph sees the "Whole Picture" 🔍, allowing it to fuse operations and eliminate memory bottlenecks that standard libraries can't see.
* **🎭 The Custom Op Pattern:** You can write a kernel in Mojo (like our `conv1d`) and register it as a "Custom Op" that looks and feels like a native PyTorch or TensorFlow operation.
* **🚫📋 Zero-Copy Memory:** It manages the GPU buffers (`DeviceContextPtr`) for you, ensuring data stays on the hardware and doesn't constantly round-trip back to the CPU.
* **🌐⚙️ Hardware Agnostic:** You write the Graph once; the compiler decides whether to execute your Mojo kernels on the CPU 🖥️ or the GPU 🎮 based on the target you provide.

### **🎯⚖️ The Choice: Direct Interop vs. MAX Graph**

| Feature | 🔄 Direct Interop | 📊 MAX Graph |
|---------|-------------------|---------------|
| **⚡ Execution** | **🐌 Step-by-step (Sequential):** Executes operations one by one as they are called in the Python code. | **🧠 Symbolic (Optimized Plan):** Compiles the entire logic into a unified execution plan before running. |
| **🔀 Parallelism** | **🔒 Limited by Python's GIL:** Subject to the constraints of the Python Global Interpreter Lock. | **🚀 Native Multi-threaded:** Runs on a dedicated high-performance runtime, bypassing Python's threading limits. |
| **🎮 GPU Optimization** | **✋ Manual Stream Management:** Requires the developer to manually handle GPU synchronization and streams. | **🤖 Automatic Scheduling:** The Graph engine intelligently schedules kernels to maximize hardware utilization. |
| **💾 Memory** | **📋 Often requires copies:** Moving data between Python and Mojo frequently involves memory overhead. | **🚫📋 Pure Zero-Copy:** Tensors are passed directly through the graph without unnecessary movement or duplication. |

# Lets check out the code...


````python

from pathlib import Path
import os
import numpy as np
from max.driver import CPU, Accelerator, Tensor, accelerator_count
from max.dtype import DType
from max.engine import InferenceSession
from max.graph import DeviceRef, Graph, TensorType, ops
````

**Step 1**  Location

````python
 
mojo_kernels = Path(os.getcwd()) / "kernels"
mojo_kernels
````

````python
vector_width = 10
dtype = DType.float32

# Place the graph on a GPU, if available. Fall back to CPU if not.
device = CPU() if accelerator_count() == 0 else Accelerator()
device
````

**Step 2** Create Max Tensors

````python

# Fill input matrices with random values using PYTHON NUMPY
lhs_values = np.random.uniform(size=(vector_width)).astype(np.float32)
rhs_values = np.random.uniform(size=(vector_width)).astype(np.float32)

# Create max tensors from this, and move them to the accelerator.
lhs_tensor = Tensor.from_numpy(lhs_values).to(device)
rhs_tensor = Tensor.from_numpy(rhs_values).to(device)
````

## **Building your first computation graph**

````python

######## User : I want to execute Mojo from Python
######## Graph : You want to execute mojo.  No problem. What type of data are you sending? 
######## Graph : Where to find mojo code? 
######## User : Let me tell you. 

with Graph(
    "my_first_max_graph" ,   #name

    input_types=[            #Input types 
        TensorType(
            dtype,
            shape=[vector_width],
            device=DeviceRef.from_device(device),
        ),
        TensorType(
            dtype,
            shape=[vector_width],
            device=DeviceRef.from_device(device),
        ),
    ],

    custom_extensions=[mojo_kernels],    # Mojo code to find
    
) as graph:


##################################### Within the context of graph ###########################################
######## Graph : Hmm let me think

    ### what are my inputs?
    lhs, rhs = graph.inputs  


    ### what custom operation should I execute? 
    ### This better be in the 'custom_extensions' folder
    output = ops.custom(
        name="vector_addition",
        device=DeviceRef.from_device(device),  # ctx
        values=[lhs, rhs],    # inputs 2 of them
        out_types=[          # output one of them
            TensorType(
                dtype=lhs.tensor.dtype,
                shape=lhs.tensor.shape,
                device=DeviceRef.from_device(device),
            )
        ],
        parameters={
            "output_size": vector_width,
            "dtype": DType.float32
        },
    )[0].tensor

    ### After completion, what should i return?
    graph.output(output)

#############################################################################################################
````

## **Execute graph**

````python

 # Set up an inference session for running the graph.
session = InferenceSession(
        devices=[device],
    )

# Compile the graph.
model = session.load(graph)

# Perform the calculation on the target device.
result = model.execute(lhs_tensor, rhs_tensor)[0]

# Copy values back to the CPU to be read.)
result = result.to(CPU())

print("Left-hand-side values:")
print(lhs_values)
print()

print("Right-hand-side values:")
print(rhs_values)
print()

print("Graph result:")
print(result.to_numpy())
print()

````

## Lets try convolution!

````python

from max.driver import CPU, Accelerator, Device, Tensor, accelerator_count
from max.dtype import DType
from max.engine import InferenceSession
from max.graph import DeviceRef, Graph, TensorType, ops
from typing import Optional
from pathlib import Path
import numpy as np
````

````python
INPUT_SIZE = 15
KERNEL_SIZE = 4
device = Accelerator()
````

````python


# Create Numpy Arrays
input_array = np.arange(INPUT_SIZE, dtype=np.float32)
kernel_array = np.arange(KERNEL_SIZE, dtype=np.float32)

# Convert to MAX Graph tensors
input_max_tensor = Tensor.from_numpy(input_array).to(device)
filter_max_tensor = Tensor.from_numpy(kernel_array).to(device)

````

````python
# Load our custom operations
mojo_kernels = Path(os.getcwd()) / "kernels"  # for python script : Path(__file__).parent / "kernels"

##################################### Graph Entrance ###########################################
with Graph(
    "conv_1d_graph",
    input_types=[
        TensorType(
            DType.float32,
            shape=input_max_tensor.shape,
            device=DeviceRef.from_device(device),
        ),
        TensorType(
            DType.float32,
            shape=filter_max_tensor.shape,
            device=DeviceRef.from_device(device),
        ),
    ],
    custom_extensions=[mojo_kernels],

) as graph_conv:
    

    ##################################### Within the context of graph ###########################################

    input_value, kernel_value = graph_conv.inputs

    output = ops.custom(
        name="conv1d",  # <---- your mojo kernel!
        values=[input_value, kernel_value],
        device=DeviceRef.from_device(device),
        out_types=[
            TensorType(
                dtype=input_value.tensor.dtype,
                shape=input_value.tensor.shape,
                device=DeviceRef.from_device(device),
            )
        ],
        parameters={
            "input_size": input_max_tensor.shape[0],
            "conv_size": filter_max_tensor.shape[0],
            "dtype": DType.float32,
        },
    )[0].tensor

    graph_conv.output(output)
````

````python
# load model
model_conv = session.load(graph_conv)
# Execute
result_conv = model_conv.execute(input_max_tensor, filter_max_tensor)[0]
````
